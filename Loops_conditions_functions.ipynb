# Control FLow Tools

```python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int

def describe_point(p):
    match p:
        case Point(0, 0):
            return "Origin"
        case Point(0, y):
            return f"Y-axis at {y}"
        case Point(x, 0):
            return f"X-axis at {x}"
        case Point(x, y) if x == y:
            return "Diagonal line"
        case _:
            return "Somewhere else"

print(describe_point(Point(0, 3)))
```
