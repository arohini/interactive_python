# Decorators

### Decorators are flexible way to modify or extend behavior of functions or methods, without changing their actual code

# Basic Decorator

```python
import time
def custom_timer(func): # main function
    def time_it(): # inner function
        start_time = time.time()
        print(f"start time is {start_time}")
        func()
        end_time = time.time()
        print(f"End time is {end_time}")
        total_time_taken = end_time - start_time
        print(f"Total time taken is {total_time_taken}")
    return time_it
        
```

```python
@custom_timer # decorator name
def add_numbers():
    a = 10
    b = 20
    sum = a + b
    print(f"sum of {a} + {b} is {sum}")
    return sum

add_numbers()
```

# Multiple Decorators

### Decorator applied in the order of bottom to top
### Executes in the order of top to bottom

```python
# Decorator 1: Convert result to uppercase
def to_uppercase(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result.upper()
    return wrapper

# Decorator 2: Add a greeting before the result
def add_greeting(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return f"Hello, {result}"
    return wrapper

# Apply both decorators
@to_uppercase
@add_greeting
def get_name():
    return "Learning New Skill Is New Fantasy"

# Test
print(get_name())
```

# Argument Passing

### Know fixed num of arguments to decorators

```python
def log_call(func):
    def wrapper(name):
        print(f"[LOG] Calling {func.__name__} with argument: {name}")
        result = func(name)
        print(f"[LOG] {func.__name__} returned===========>>>>>>>: {result}")
        return result
    return wrapper

@log_call
def get_car_features(features):
    uc = ' '. join(features).upper()
    return f"common features, {uc}!"

# Test
print(get_car_features(["ABS", "Airbag", "Engine", "Brakes"]))
```

### Decorator for Unknown arguments

```python
import functools
import time

# --- Decorator 1: Authentication ---
def authenticate_user(func):
    @functools.wraps(func) ### Technique to preserve function metadata
    def wrapper(*args, **kwargs): # To accomodate Unknown arguments
        user = kwargs.get('user')
        if not user or not user.get('authenticated'):
            raise PermissionError("User not authenticated!")
        print(f"[AUTH] User '{user['name']}' authenticated.")
        return func(*args, **kwargs)
    return wrapper


# --- Decorator 2: Logging ---
def log_execution(func):
    @functools.wraps(func) ### Technique to preserve function metadata
    def wrapper(*args, **kwargs): # To accomodate Unknown arguments
        start = time.time()
        print(f"[LOG] Starting '{func.__name__}'...")
        result = func(*args, **kwargs)
        duration = time.time() - start
        print(f"[LOG] Finished '{func.__name__}' in {duration:.3f}s.")
        return result
    return wrapper



# --- Combining Decorators ---
@authenticate_user
@log_execution
def get_user_data(*, user):
    """Simulate fetching user data (e.g., from a database)."""
    time.sleep(1)  # Simulate delay
    return {"id": 1, "name": user['name'], "role": "admin"}


# --- Testing ---
user = {"name": "Rohini", "authenticated": True}

print("\nFirst call:")
print(get_user_data(user=user))

```

### Technique to preserve function metadata

```python
import functools

def changecase(func):
  @functools.wraps(func)
  def myinner():
    print("Have a great day!!")
    return func().upper()
  return myinner

@changecase
def myfunction():
  print("Have a great day!")
  return "Have a great day!"
myfunction()
print(myfunction.__name__)
```
