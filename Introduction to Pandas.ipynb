## Pandas is built on top of the NumPy library and provides efficient data structures for handling structured data.

```python
# Import the Pandas library, conventionally aliased as 'pd'
import pandas as pd
import numpy as np

print(f"Pandas Version: {pd.__version__}")
```

### Pandas primarily uses two data structures: Series and DataFrame.

### 2.1. Series (One-dimensional labeled array)

### A Series is like a single column in a spreadsheet. It holds a sequence of values and an associated array of labels (the index).

```python
# Creating a Series from a Python list
data_list = [10, 20, 30, 40, 50]
s = pd.Series(data_list)
print("--- Series from list ---")
print(s)

# Creating a Series with custom labels (index)
s_labeled = pd.Series([1.5, 2.5, 3.5], index=['a', 'b', 'c'])
print("\n--- Series with custom index ---")
print(s_labeled)

# Accessing elements
print(f"\nValue at index 2: {s[2]}")
print(f"Value at label 'b': {s_labeled['b']}")
```

## 2.2. DataFrame (Two-dimensional labeled data structure)

## A DataFrame is the most common Pandas object, representing a table of data with rows and columns (like a spreadsheet or SQL table).

```python
# Creating a DataFrame from a dictionary of lists
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', None],
    'Age': [25, 30, 35, 40, 0],
    'City': ['New York', 'London', None, 'Tokyo', None]
}
df = pd.DataFrame(data)

print("--- Initial DataFrame ---")
print(df)
print("\n--- DataFrame Info ---")
df.info()
```

### Inspection

```python
# Display the first 5 rows (default)
print("--- Head (first 3 rows) ---")
print(df.head(3))

# Display the last 5 rows (default)
print("\n--- Tail (last 2 rows) ---")
print(df.tail(2))

# Get summary statistics for numerical columns
print("\n--- Describe (summary statistics) ---")
print(df.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns


# Select only numeric questions (1–5 scale)
numeric_cols = [
    "9. How often do you find yourself using Social media without a specific purpose?",
    "10. How often do you get distracted by Social media when you are busy doing something?",
    "11. Do you feel restless if you haven't used Social media in a while?",
    "12. On a scale of 1 to 5, how easily distracted are you?",
    "13. On a scale of 1 to 5, how much are you bothered by worries?",
    "14. Do you find it difficult to concentrate on things?",
    "15. On a scale of 1-5, how often do you compare yourself to other successful people through the use of social media?",
    "16. Following the previous question, how do you feel about these comparisons, generally speaking?",
    "17. How often do you look to seek validation from features of social media?",
    "18. How often do you feel depressed or down?",
    "19. On a scale of 1 to 5, how frequently does your interest in daily activities fluctuate?",
    "20. On a scale of 1 to 5, how often do you face issues regarding sleep?"
]

df_numeric = sm_data[numeric_cols]

# --- 1. Average scores ---
print("\nAverage Scores:")
print(df_numeric.mean())

# Bar plot of average scores
df_numeric.mean().plot(kind='bar', figsize=(12,5), color='skyblue')
plt.title("Average Scores for Survey Questions")
plt.ylabel("Average Score (1-5)")
plt.show()
```

```python
def map_usage(x):
    x = str(x).lower()
    if "less than 1" in x:
        return "Low"
    elif "1" in x or "2" in x or "between" in x:
        return "Medium"
    elif "more" in x or "3" in x:
        return "High"
    return "Unknown"

sm_data["Usage_Level"] = sm_data["8. What is the average time you spend on social media every day?"].apply(map_usage)
print(sm_data["Usage_Level"].value_counts())
```
