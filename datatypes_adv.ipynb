# Named Tuples

## Creation
### You create a namedtuple by calling the namedtuple() function, 
### providing a name for the new tuple type and a list or string of field names.

```python
from collections import namedtuple

# Using a list of strings for field names
MongoClient = namedtuple('MongoClient', ['host', 'port', 'db_name', 'collection_name'])

MongoClient_new = namedtuple('MongoClient_new', ['host', 'port', 'db_name', 'collection_name'])

# Using a single string with space-separated field names
MongoClient = namedtuple('MongoClient', 'host port db_name collection_name')
```

# Instantiation
### Once defined, you can create instances of your namedtuple like any other class, passing values for the defined fields.

```python
mc = MongoClient(host='localhost', port='27107', db_name='python', collection_name='assignments')
mc_c = MongoClient('localhost', 27107, 'python', 'assignments')
```

# Accessing Elements:
### The primary advantage of namedtuple is accessing elements by name using dot notation, in addition to traditional index-based access.

```python
print(mc.host)  
print(mc.port)  
print(mc.db_name) 
print(mc.collection_name) 

mc[0]
```
